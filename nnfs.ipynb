```python
import numpy as np
import matplotlib.pyplot as plt
import struct 
import os
```

```python
data_path = "data"
print (os.listdir(data_path))
```

```python
image_file = os.path.join(data_path, "train-images.idx3-ubyte")
label_file = os.path.join(data_path, "train-labels.idx1-ubyte")
print("Image file exists:", os.path.exists(image_file))
print("Label file exists:", os.path.exists(label_file))
```

```python
with open(image_file, "rb") as file:
    header = file.read(16)
    print(header)
```

```python
magic_number, num_images, rows, cols = struct.unpack(">IIII", header)

print("Magic Number :", magic_number)
print("Number of Images :", num_images)
print("Rows :", rows)
print("Columns :", cols)
```

```python
with open(image_file, "rb") as file:

    file.read(16)

    first_image = file.read(784)

    print(type(first_image))
    print(len(first_image))
    print(first_image)
```

```python
first_image = np.frombuffer(first_image, dtype = np.uint8)
print(first_image.shape)
```

```python
first_image = first_image.reshape(28, 28)
print(first_image.shape)
```

```python
plt.imshow(first_image, cmap="gray")
```

```python
with open(image_file, "rb") as file:
    file.read(800)
    second_img = file.read(784)
second_img = np.frombuffer(second_img, dtype=np.uint8)
second_img = second_img.reshape(28, 28)
plt.imshow(second_img, cmap="gray")
```

```python
def see_img(n):
    with open(image_file, "rb") as file:
        file.read(16+((n-1)*784))
        your_img = file.read(784)
    your_img = np.frombuffer(your_img, dtype=np.uint8)
    your_img = your_img.reshape(28, 28)
    plt.imshow(your_img, cmap="gray")
see_img(2)
```
